```python
import time
import random
import re
from urllib.parse import urljoin
import pandas as pd
from bs4 import BeautifulSoup
import requests

# Configuration
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/150.0.0.0 Safari/537.36"
    ),
    "Accept-Language": "en-US,en;q=0.9",
}
BASE_URL = "https://www.carwale.com"

# Dictionary to hold separate lists of URLs for each brand
brand_cars_map = {}

# ==============================================================================
# STAGES 1 & 2: EXTRACT AND FILTER LIVE NEW CAR MODEL URLS BY BRAND
# ==============================================================================
print("--- STAGES 1 & 2: Extracting Isolated Live Car Model URLs ---")

# Loop through your brand_urls array (ensure this list is active in memory)
for brand_url in brand_urls:
    brand_slug = brand_url.split("/")[-2].replace("-cars", "").lower()
    brand_name = brand_slug.capitalize()
    
    print(f"Filtering active models for brand: {brand_name}...")
    
    try:
        response = requests.get(brand_url, headers=headers, timeout=10)
        if response.status_code != 200:
            continue
            
        soup = BeautifulSoup(response.text, "lxml")
        brand_cars_map[brand_name] = []
        
        # Look for all anchor tags pointing to individual model pages
        model_links = soup.find_all("a", href=True)
        for link in model_links:
            href = link.get("href").strip()
            
            # Form full url path for validation checks
            full_model_url = urljoin(BASE_URL, href)
            
            # 1. Broad Validation Check: Ensure it belongs strictly to this brand's sub-path
            if f"/{brand_slug}-cars/" in href.lower():
                
                # 2. Year-Range Exclusion: Drop legacy generations (e.g., wrangler-2021-2024)
                if re.search(r"\d{4}-\d{4}", href):
                    continue
                    
                # 3. Text Keyword Exclusion: Drop expert reviews, videos, upcoming, and media layout clutter
                noise_keywords = [
                    "compare", "upcoming", "old-generation", "used", "history", 
                    "discontinued", "price-in-", "images", "colors", "videos",
                    "expert-reviews", "reviews", "news", "features"
                ]
                if any(keyword in href.lower() for keyword in noise_keywords):
                    continue
                
                # 4. Filter check to make sure it doesn't just point back to the parent page itself
                if href.lower().rstrip('/') == f"/{brand_slug}-cars":
                    continue
                
                # 5. Duplicate Protection
                if full_model_url not in brand_cars_map[brand_name]:
                    brand_cars_map[brand_name].append(full_model_url)
                    
        print(f"-> Verified {len(brand_cars_map[brand_name])} active new model URLs for {brand_name}.")
        time.sleep(random.uniform(1, 1.5))  
        
    except Exception as e:
        print(f"Error filtering models for brand {brand_name}: {e}")

# ==============================================================================
# STAGE 3: DEEP DATA EXTRACTION
# ==============================================================================
print("\n--- STAGE 3: Extracting Specifications from Live URLs ---")
final_car_data = []

for brand_name, car_urls in brand_cars_map.items():
    for car_url in car_urls:
        print(f"Scraping detailed specifications from: {car_url}")
        
        try:
            response = requests.get(car_url, headers=headers, timeout=10)
            if response.status_code != 200:
                continue
                
            soup = BeautifulSoup(response.text, "lxml")
            
            # 1. Extract Car Name from h1
            title_element = soup.find("h1")
            car_name = title_element.text.replace(brand_name, "").strip() if title_element else "Unknown"
            
            # 2. Extract User Rating using verified parent container class pattern
            rating_element = soup.find("p", class_=re.compile(r"o-iY.*o-jq"))
            user_rating = rating_element.text.strip() if rating_element else "Not Rated"
            if user_rating != "Not Rated" and not user_rating.endswith("/5"):
                user_rating = f"{user_rating}/5"
                
            # 3. Handle Variant Layout Blocks
            variant_blocks = soup.find_all("div", data_version_id=True) or soup.find_all("div", class_=re.compile(r"sc-jq0284"))
            
            # Fallback directly to text container parsing if structures shift
            if not variant_blocks:
                variant_blocks = soup.find_all("div")

            for block in variant_blocks:
                # Get inline text context
                block_text = block.get_text(separator=" | ")
                
                # Verify rows containing pricing metadata tokens
                if "Rs." not in block_text or "|" not in block_text:
                    continue
                    
                # Parse Price
                price_match = re.search(r"Rs\.\s*\d+(?:\.\d+)?\s*(?:Lakh|Crore)", block_text)
                price = price_match.group(0) if price_match else "On Request"
                
                fuel_type = "Unknown"
                transmission = "Unknown"
                engine = "Unknown"
                
                # Identify Fuel Type
                if "Petrol" in block_text:
                    fuel_type = "Petrol"
                elif "Diesel" in block_text:
                    fuel_type = "Diesel"
                elif "CNG" in block_text:
                    fuel_type = "CNG"
                elif "Electric" in block_text:
                    fuel_type = "Electric"
                    
                # Identify Transmission
                if "Manual" in block_text:
                    transmission = "Manual"
                elif "Automatic" in block_text or "DCT" in block_text or "AMT" in block_text:
                    transmission = "Automatic"
                    
                # Identify Engine Capacity / Battery Power
                engine_match = re.search(r"(\d+\.\d+L(?::\s*Turbo)?|\d+\s*cc|\d+\s*kWh)", block_text)
                if engine_match:
                    engine = engine_match.group(0)
                elif fuel_type == "Electric":
                    engine = "Electric Motor"
                    
                # 4. Extract Mileage / Range using text string patterns
                mileage = "N/A"
                mileage_match = re.search(r"(\d+(?:\.\d+)?\s*(?:kmpl|km/kg|km/charge))", block_text)
                if mileage_match:
                    mileage = mileage_match.group(0)
                    
                # Structure entry
                record = {
                    "Car Name": car_name,
                    "Brand": brand_name,
                    "Price": price,
                    "Fuel Type": fuel_type,
                    "Transmission": transmission,
                    "Engine": engine,
                    "Mileage": mileage,
                    "User Rating": user_rating
                }
                
                # Avoid inserting repeating exact row layouts
                if record not in final_car_data:
                    final_car_data.append(record)
                
            time.sleep(random.uniform(0.3, 0.7))
            
        except Exception as e:
            print(f"Error extracting deep data from {car_url}: {e}")

# ==============================================================================
# FINAL RESULTS STRUCTURING
# ==============================================================================
df = pd.DataFrame(final_car_data)
print(f"\nExtraction Complete! Found {len(df)} rows.")
if not df.empty:
    print(df.head())
```

```python
import time
import random
import re
from urllib.parse import urljoin
import pandas as pd
from bs4 import BeautifulSoup
import requests

# Configuration
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/150.0.0.0 Safari/537.36"
    ),
    "Accept-Language": "en-US,en;q=0.9",
}
BASE_URL = "https://www.carwale.com"

# Dictionary to hold metadata dictionaries for each brand
# Format: { Brand: [ {"url": URL, "rating": RATING}, ... ] }
brand_cars_map = {}

# ==============================================================================
# STAGES 1 & 2: EXTRACT AND FILTER LIVE NEW CAR MODEL URLS + RATINGS
# ==============================================================================
print("--- STAGES 1 & 2: Extracting Isolated Live Car Model URLs & Ratings ---")

for brand_url in brand_urls:
    brand_slug = brand_url.split("/")[-2].replace("-cars", "").lower()
    brand_name = brand_slug.capitalize()
    
    print(f"Filtering active models for brand: {brand_name}...")
    
    try:
        response = requests.get(brand_url, headers=headers, timeout=10)
        if response.status_code != 200:
            continue
            
        soup = BeautifulSoup(response.text, "lxml")
        brand_cars_map[brand_name] = []
        
        # Scrape all individual model container blocks if available, or anchor loops
        model_links = soup.find_all("a", href=True)
        
        # Keep track of unique URLs per brand to prevent duplication
        seen_urls = set()
        
        for link in model_links:
            href = link.get("href").strip()
            full_model_url = urljoin(BASE_URL, href)
            
            # 1. Path Filtering: Must match the brand's main model listing pattern
            if f"/{brand_slug}-cars/" in href.lower():
                
                # 2. Exclude Legacy Models & Clutter Keywords
                if re.search(r"\d{4}-\d{4}", href):
                    continue
                noise_keywords = [
                    "compare", "upcoming", "old-generation", "used", "history", 
                    "discontinued", "price-in-", "images", "colors", "videos",
                    "expert-reviews", "reviews", "news", "features"
                ]
                if any(keyword in href.lower() for keyword in noise_keywords):
                    continue
                if href.lower().rstrip('/') == f"/{brand_slug}-cars":
                    continue
                
                if full_model_url not in seen_urls:
                    seen_urls.add(full_model_url)
                    
                    # --- COMPONENT FIX: Extract User Rating right here from the parent context ---
                    user_rating = "Not Rated"
                    # Look globally around the anchor tag context for the custom class rating badge
                    parent_container = link.find_parent(["div", "li"])
                    if parent_container:
                        rating_el = parent_container.find("p", class_=re.compile(r"o-iY.*o-jq")) or parent_container.find(text=re.compile(r"^\d\.\d$"))
                        if rating_el:
                            raw_val = rating_el.get_text().strip()
                            # Clean rating string values
                            user_rating = f"{raw_val}/5" if "/5" not in raw_val else raw_val
                    
                    brand_cars_map[brand_name].append({
                        "url": full_model_url,
                        "rating": user_rating
                    })
                    
        print(f"-> Verified {len(brand_cars_map[brand_name])} active new models for {brand_name}.")
        time.sleep(random.uniform(1, 1.5))  
        
    except Exception as e:
        print(f"Error filtering models for brand {brand_name}: {e}")

# ==============================================================================
# STAGE 3: CLEAN DEEP DATA EXTRACTION
# ==============================================================================
print("\n--- STAGE 3: Extracting Specifications from Live URLs ---")
final_car_data = []

for brand_name, car_items in brand_cars_map.items():
    for item in car_items:
        car_url = item["url"]
        user_rating = item["rating"]
        
        print(f"Scraping valid specs from: {car_url}")
        
        try:
            response = requests.get(car_url, headers=headers, timeout=10)
            if response.status_code != 200:
                continue
                
            soup = BeautifulSoup(response.text, "lxml")
            
            # 1. Extract Car Name
            title_element = soup.find("h1")
            car_name = title_element.text.replace(brand_name, "").strip() if title_element else "Unknown"
            
            # 2. Extract Variant Info blocks
            variant_blocks = soup.find_all("div", data_version_id=True) or soup.find_all("div", class_=re.compile(r"sc-jq0284"))
            if not variant_blocks:
                variant_blocks = soup.find_all("div")

            for block in variant_blocks:
                block_text = block.get_text(separator=" | ")
                
                # --- COMPONENT FIX: Strict layout validation ---
                # Drop structural boxes that do not contain a valid Fuel Type string
                if not any(f in block_text for f in ["Petrol", "Diesel", "CNG", "Electric"]):
                    continue
                if "Rs." not in block_text:
                    continue
                    
                # Parse Price
                price_match = re.search(r"Rs\.\s*\d+(?:\.\d+)?\s*(?:Lakh|Crore)", block_text)
                price = price_match.group(0) if price_match else "On Request"
                
                fuel_type = "Unknown"
                transmission = "Unknown"
                engine = "Unknown"
                
                # Map Fuel Type
                if "Petrol" in block_text: fuel_type = "Petrol"
                elif "Diesel" in block_text: fuel_type = "Diesel"
                elif "CNG" in block_text: fuel_type = "CNG"
                elif "Electric" in block_text: fuel_type = "Electric"
                    
                # Map Transmission
                if "Manual" in block_text:
                    transmission = "Manual"
                elif any(t in block_text for t in ["Automatic", "DCT", "AMT", "CVT", "Torque Converter"]):
                    transmission = "Automatic"
                    
                # Map Engine Size / Power metrics
                engine_match = re.search(r"(\d+\.\d+L(?::\s*Turbo)?|\d+\s*cc|\d+\s*kWh)", block_text)
                if engine_match:
                    engine = engine_match.group(0)
                elif fuel_type == "Electric":
                    engine = "Electric Motor"
                    
                # 4. Map Mileage metric
                mileage = "N/A"
                mileage_match = re.search(r"(\d+(?:\.\d+)?\s*(?:kmpl|km/kg|km/charge))", block_text)
                if mileage_match:
                    mileage = mileage_match.group(0)
                    
                record = {
                    "Car Name": car_name,
                    "Brand": brand_name,
                    "Price": price,
                    "Fuel Type": fuel_type,
                    "Transmission": transmission,
                    "Engine": engine,
                    "Mileage": mileage,
                    "User Rating": user_rating
                }
                
                if record not in final_car_data:
                    final_car_data.append(record)
                
            time.sleep(random.uniform(0.3, 0.6))
            
        except Exception as e:
            print(f"Error extracting deep data from {car_url}: {e}")

# Save final structured data
df = pd.DataFrame(final_car_data)
print(f"\nExtraction Complete! Formatted a clean table with {len(df)} rows.")
if not df.empty:
    print(df.head(10))
```
