```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import json
import re
import random
import time
```

```python
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36"
}
```

```python
BASE_URL = "https://www.carwale.com"
```

```python
def get_soup(url):

    response = requests.get(url, headers=HEADERS)

    if response.status_code == 200:
        return BeautifulSoup(response.text, "html.parser")

    else:
        print("Failed:", url)
        return None
```

```python
def get_brand_urls():

    soup = get_soup(BASE_URL + "/new-cars/")

    brand_urls = set()

    for tag in soup.find_all("a", href=True):

        href = tag["href"]

        # Only URLs like /tata-cars/
        if not re.fullmatch(r"/[a-z0-9-]+-cars/", href):
            continue

        brand_urls.add(BASE_URL + href)

    return sorted(brand_urls)
```

```python
brand_urls = get_brand_urls()

print("Brands:", len(brand_urls))

brand_urls
```

```python
def get_car_urls(brand_url):

    soup = get_soup(brand_url)

    car_urls = set()

    # Example:
    # https://www.carwale.com/tata-cars/
    # --> tata-cars
    brand_slug = brand_url.rstrip("/").split("/")[-1]

    bad_words = [
        "price-in",
        "expert-reviews",
        "images",
        "videos",
        "news",
        "old-generation",
        "prime",
        "jtp",
        "genx",
        "2011",
        "2012",
        "2013",
        "2014",
        "2015",
        "2016",
        "2017",
        "2018",
        "2019",
        "2020",
        "2021",
        "2022",
        "2023",
        "2024",
        "1991"
    ]

    for tag in soup.find_all("a", href=True):

        href = tag["href"]

        # Must start with this brand only
        if not href.startswith(f"/{brand_slug}/"):
            continue

        # Only model URLs
        if href.count("/") != 3:
            continue

        if "#" in href or "?" in href:
            continue

        if any(word in href.lower() for word in bad_words):
            continue

        car_urls.add(BASE_URL + href)

    return sorted(car_urls)
```

```python
tata_urls = get_car_urls("https://www.carwale.com/tata-cars/")

print(len(tata_urls))

for u in tata_urls:
    print(u)
```

```python
all_car_urls = []

for brand in brand_urls:

    print(f"Scraping: {brand}")

    try:
        urls = get_car_urls(brand)

        all_car_urls.extend(urls)

    except Exception as e:
        print("Error:", e)

    time.sleep(random.uniform(1, 2))
```

```python
all_car_urls = sorted(set(all_car_urls))

print("Total Cars:", len(all_car_urls))
```

```python
def scrape_car(url):

    response = requests.get(url, headers=HEADERS)

    if response.status_code != 200:
        return None

    soup = BeautifulSoup(response.text, "html.parser")

    # -----------------------------
    # Find JSON-LD script
    # -----------------------------
    ld_json = None

    for script in soup.find_all("script", type="application/ld+json"):

        try:
            data = json.loads(script.string)

            if "@graph" in data:

                for item in data["@graph"]:

                    if item.get("@type") == "Car":
                        ld_json = item
                        break

        except:
            pass

    if ld_json is None:
        return None

    # -----------------------------
    # Extract Price
    # -----------------------------
    price = np.nan

    if "offers" in ld_json:

        offer = ld_json["offers"]

        if isinstance(offer, dict):
            price = offer.get("price", np.nan)

    # -----------------------------
    # Extract Brand
    # -----------------------------
    brand = np.nan

    if "brand" in ld_json:

        if isinstance(ld_json["brand"], dict):
            brand = ld_json["brand"].get("name", np.nan)
        else:
            brand = ld_json["brand"]

    # -----------------------------
    # Extract Specifications
    # -----------------------------
    description = ld_json.get("description", "")

    fuel = np.nan
    transmission = np.nan
    engine = np.nan
    mileage = np.nan
    seating = np.nan

    # Fuel Type
    fuel_match = re.search(
        r"(Petrol|Diesel|CNG|Electric|Hybrid)",
        description,
        re.I
    )

    if fuel_match:
        fuel = fuel_match.group(1).title()

    # Transmission
    transmission_match = re.search(
        r"(Manual|Automatic)",
        description,
        re.I
    )

    if transmission_match:
        transmission = transmission_match.group(1).title()

    # Engine Capacity
    engine_match = re.search(
        r"(\d{3,4})\s*to\s*(\d{3,4})\s*cc",
        description,
        re.I
    )

    if engine_match:
        # Example: 1199 to 1497 cc -> take 1199
        engine = engine_match.group(1)

    else:
        engine_match = re.search(
            r"(\d{3,4})\s*cc",
            description,
            re.I
        )

        if engine_match:
            engine = engine_match.group(1)

    # Mileage
    mileage_match = re.search(
        r"mileage of\s*([\d\.]+)",
        description,
        re.I
    )

    if mileage_match:
        mileage = mileage_match.group(1)

    # Seating Capacity
    seat_match = re.search(
        r"(\d)\s*seater",
        description,
        re.I
    )

    if seat_match:
        seating = seat_match.group(1)

    # -----------------------------
    # User Rating
    # -----------------------------
    rating = np.nan

    if "aggregateRating" in ld_json:

        rating = ld_json["aggregateRating"].get("ratingValue", np.nan)

    # -----------------------------
    # Return dictionary
    # -----------------------------
    return {
        "Car Name": ld_json.get("name", np.nan),
        "Brand": brand,
        "Price": price,
        "Fuel Type": fuel,
        "Transmission": transmission,
        "Engine (cc)": engine,
        "Mileage": mileage,
        "Seating Capacity": seating,
        "User Rating": rating,
        "URL": url
    }
```

```python
all_cars = []

for i, url in enumerate(all_car_urls, start=1):

    print(f"{i}/{len(all_car_urls)} : {url}")

    try:

        data = scrape_car(url)

        if data:
            all_cars.append(data)

    except Exception as e:

        print("Error:", e)

    time.sleep(random.uniform(1, 2))
```

```python
df = pd.DataFrame(all_cars)

print(df.shape)

df.head()
```
